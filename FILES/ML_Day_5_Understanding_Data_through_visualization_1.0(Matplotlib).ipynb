#### `Subplot`
* Syntax E.g - fig,ax=plt.subplots(nrows=1,ncols=2,figsize=(7,5))

```python
height=[160,145,155,120,130,176,175]
weight=[67,56,45,55,76,55,78]
height.sort()
weight.sort()
calories=[50,55,58,60,75,78,90]

fig, ax = plt.subplots(nrows=1,ncols=2,figsize=(10,6))
ax[0].plot(weight,height,label ="weight vs height", marker = "*", color = "r")
ax[0].set_xlabel("Weight")
ax[0].set_ylabel("Height")
ax[0].set_title("weight vs Height")
ax[1].plot(weight,calories,label ="weight vs calories" , marker = "o")
ax[1].set_xlabel("Weight")
ax[1].set_ylabel("Calories")
ax[1].set_title("weight vs calories")
ax[0].legend()
ax[1].legend()
plt.show()
```

#### `Histogram`
* What it shows: Frequency distribution of a continuous variable by dividing data into bins.
* Understand the shape (skewness, modality) of a numerical variable.
* Detect outliers or gaps.
* Compare against a theoretical distribution (e.g., normal).
* syntax plt.hist(x, bins=n, color='blue', edgecolor='black')

#### `Bar Chart (for counts)`
* What it shows: Frequency of each category for a categorical variable.
* Display counts of discrete groups (e.g., survey responses, product categories).
* 
Compare the size of categories visually.
* plt.bar(categories, counts, color='green')


```python
## Plot bar of the student across course
```

```python
course = ["BCA", "BBA", "BSC_AI_DS"]
no_Student = [200, 120, 127]
plt.bar(course, no_Student,color = ["r","g","b"])
plt.xlabel("Courses")
plt.ylabel("No of Student")
plt.title("Student Distribution Across Courses")
plt.show()
```

#### `Pie Chart`
* What it shows: Proportional makeup of a categorical variable as slices of a circle.
* Emphasise part‑to‑whole relationships when there are few categories (≤5).* 
Caution: Hard to compare angles; often replaced by bar charts.
* plt.pie(sizesexplode=explode,, labels=labels, autopct='%1.1f%%', startangle=9,shadow=True0)

```python
course = ["BCA", "BBA", "BSC_AI_DS", "B.Tech"]
no_Student = [200, 120, 127,340]
explode = [0.04,0,0.03,0.14]
plt.figure(figsize=(7,4))
plt.pie(no_Student, explode=explode,labels=course, autopct="%1.1f%%", startangle=100, shadow=True,rotatelabels=120,labeldistance=1.1)
plt.title("Distribution of Student across Courses")
plt.savefig("Distribution of Student across Courses", dpi=400)
plt.show()
```
